## 4. Two first plots

```python
fig, ax = plt.subplots(figsize=(7, 4))
for sp, g in df.groupby("species"):
    ax.hist(g["flipper_length_mm"].dropna(), bins=20, alpha=0.6, label=sp)
ax.set_xlabel("Flipper length (mm)"); ax.set_ylabel("Number of penguins")
ax.set_title("Flipper length by species"); ax.legend(); plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
for sp, g in df.groupby("species"):
    ax.scatter(g["bill_length_mm"], g["bill_depth_mm"], s=18, alpha=0.7, label=sp)
ax.set_xlabel("Bill length (mm)"); ax.set_ylabel("Bill depth (mm)"); ax.legend(); plt.show()
print("Overall r =", round(df["bill_length_mm"].corr(df["bill_depth_mm"]), 2))
print(df.groupby("species")[["bill_length_mm", "bill_depth_mm"]].corr().xs("bill_length_mm", level=1)["bill_depth_mm"].round(2))
```

## 5. File formats: same data, three files

```python
import os, time
df.to_csv("penguins.csv", index=False)
df.to_json("penguins.json", orient="records", indent=1)
df.to_parquet("penguins.parquet", index=False)   # needs pyarrow (pre-installed in Colab)

rows = []
for f, reader in [("penguins.csv", pd.read_csv), ("penguins.json", pd.read_json), ("penguins.parquet", pd.read_parquet)]:
    t0 = time.perf_counter()
    for _ in range(20):
        back = reader(f)
    rows.append({"file": f, "size_kB": round(os.path.getsize(f) / 1024, 1),
                 "read_ms": round((time.perf_counter() - t0) / 20 * 1000, 2)})
pd.DataFrame(rows)
```
